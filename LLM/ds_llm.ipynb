```python
# 调用本地 Ollama 部署的 qwen2.5:3b 模型
from openai import OpenAI

# Ollama 默认端口 11434，OpenAI 兼容端点路径为 /v1
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"  # Ollama 本地不需要真实 API key，随便填即可
)

response = client.chat.completions.create(
    model="qwen2.5:3b",
    messages=[
        {"role": "system", "content": "你是一个经验丰富的学生管理系统助手，帮助处理学生、教师、班级、成绩等相关事务。"},
        {"role": "user", "content": "你好，请简单介绍一下你自己，并说明你能在学生管理系统中帮我做什么？"}
    ],
    stream=False
)

print(response.choices[0].message.content)
```

```python
# 流式返回
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # 从 .env 读取密钥

client = OpenAI(
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL", "https://api.deepseek.com"))

messages=[
        {"role": "system", "content": "你是一个经验丰富的工作助手，会帮助我完成一些任务"},
        {"role": "user", "content": "你好，请简单介绍一下你自己"}
    ]
response = client.chat.completions.create(
    model="deepseek-v4-flash",
    messages=messages,
    stream=True, # 是否流式返回
    reasoning_effort="high", # 思考强度
    extra_body={"thinking": {"type": "enabled"}} # 是否思考
)
for chunk in response: # 流式返回
    if chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content)
```

```python
# json格式返回
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # 从 .env 读取密钥

client = OpenAI(
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL", "https://api.deepseek.com"))

response = client.chat.completions.create(
    model="deepseek-v4-flash",
    messages=[
        {"role": "system", "content": "你是一个经验丰富的工作助手，会帮助我完成一些任务"},
        {"role": "user", "content": "你好，请简单介绍一下你自己,请以 json 格式返回结果"},
    ],
    stream=False, # 是否流式返回
    reasoning_effort="high", # 思考强度
    extra_body={"thinking": {"type": "enabled"}}, # 是否思考
    response_format={"type": "json_object"} # 返回格式
)

print(response.choices[0].message.content)
```

```python
# 对话记忆
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # 从 .env 读取密钥

client = OpenAI(
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL", "https://api.deepseek.com"))

messages=[
        {"role": "system", "content": "你是一个经验丰富的工作助手，会帮助我完成一些任务"},
        {"role": "user", "content": "你好，我是刘培坤，请简单介绍一下你自己"}
]
response = client.chat.completions.create(
    model="deepseek-v4-flash",
    messages=messages
)
messages.append(response.choices[0].message)

messages.append({"role": "user", "content": "你还记得我是谁吗？"})
response = client.chat.completions.create(
    model="deepseek-v4-flash",
    messages=messages
)
messages.append(response.choices[0].message)
print(messages)
```
